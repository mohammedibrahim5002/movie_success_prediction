```python
df.isnull().sum()
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(' ', '_')
df.columns = df.columns.str.lower()
df.columns
```

```python
df.dtypes
```

```python
df['roi'] = df['earnings'] / df['budget']
df['movie_age'] = 2026 - df['release_year']
df['total_awards'] = df['oscar_and_golden_globes_awards'] + df['oscar_and_golden_globes_nominations']
df['revenue_per_minute'] = df['box_office'] / df['running_time']
df['success'] = np.where(df['roi'] > 1, 1, 0)
df['director_avg_box_office'] = df.groupby('director')['box_office'].transform('mean')
df['actor_1_avg_box_office'] = df.groupby('actor_1')['box_office'].transform('mean')
df['actor_2_avg_box_office'] = df.groupby('actor_2')['box_office'].transform('mean')
df['actor_3_avg_box_office'] = df.groupby('actor_3')['box_office'].transform('mean')
df['cast_popularity'] = df['actor_1_avg_box_office'] + df['actor_2_avg_box_office'] + df['actor_3_avg_box_office']

df.head()
```

```python
sns.histplot(df['imdb_score'])
plt.show()

sns.scatterplot(x='budget', y='box_office', data=df)
plt.show()

genre_profit = df.groupby('genre')['earnings'].mean().sort_values(ascending=False).reset_index()
genre_profit.plot(kind='bar', x='genre', y='earnings')
plt.title('Average Earnings by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Earnings')
plt.xticks(rotation=45)
plt.show()

sns.scatterplot(x='imdb_score', y='earnings', data=df)
plt.show()

sns.scatterplot(x='running_time', y='box_office', data=df)
plt.show()

top_directors = df.groupby('director')['earnings'].mean().sort_values(ascending=False).head(10).reset_index()
top_directors.plot(kind='bar', x='director', y='earnings')
plt.title('Top 10 Directors by Average Earnings')
plt.xlabel('Director')
plt.ylabel('Average Earnings')
plt.xticks(rotation=45)
plt.show()

top_actors = df.groupby('actor_1')['earnings'].mean().sort_values(ascending=False).head(10).reset_index()
top_actors.plot(kind='bar', x='actor_1', y='earnings')
plt.title('Top 10 Main Actors by Average Earnings')
plt.xlabel('Main Actor')
plt.ylabel('Average Earnings')
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python
x = df[['running_time','genre','imdb_score','movie_age','total_awards','director_avg_box_office','actor_1_avg_box_office','actor_2_avg_box_office','actor_3_avg_box_office','cast_popularity']]
y_boxoffice = df['box_office']
y_success = df['success']
```

```python
#box_office_model

from sklearn.model_selection import train_test_split
x_train, x_test, y_train_boxoffice, y_test_boxoffice = train_test_split(x, y_boxoffice, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

numeric_cols = x_train.select_dtypes(include=['int64', 'float64']).columns
cat_cols = x_train.select_dtypes(include=['object']).columns

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
cat_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(drop='first'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('numeric', numeric_transformer, numeric_cols),
        ('categorical', cat_transformer, cat_cols)
    ]
)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
box_office_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor
     (n_estimators=300,
      max_depth=10,
      random_state=42))
])
box_office_model.fit(x_train, y_train_boxoffice)
y_pred = box_office_model.predict(x_test)
mse = mean_squared_error(y_test_boxoffice, y_pred)
r2 = r2_score(y_test_boxoffice, y_pred)
mae = mean_absolute_error(y_test_boxoffice, y_pred)
print(f'Mean Squared Error: {mse}')
print(f'R^2 Score: {r2}')
print(f'Mean Absolute Error: {mae}')
```

```python
from sklearn.metrics import classification_report
y_test_class = np.where(y_test_boxoffice > y_test_boxoffice.median(), 'Successful', 'Unsuccessful')
y_pred_class = np.where(y_pred > y_test_boxoffice.median(), 'Successful', 'Unsuccessful')
print(classification_report(y_test_class, y_pred_class))
plt.figure(figsize=(6, 6))
plt.scatter(y_test_boxoffice, y_pred,)
plt.xlabel('Actual Box Office')
plt.ylabel('Predicted Box Office')
plt.title('Actual vs Predicted Box Office')
plt.show()
```

```python
feature_importances = box_office_model.named_steps['regressor'].feature_importances_
feature_names = numeric_cols.tolist() + list(box_office_model.named_steps['preprocessor'].transformers_[1][1].named_steps['onehot'].get_feature_names_out(cat_cols))
feature_importance_df = pd.DataFrame({'Feature': feature_names, 'Importance': feature_importances})
feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', data=feature_importance_df)
plt.title('Feature Importances')
plt.xlabel('Importance')
plt.show()
```

```python
from sklearn.model_selection import cross_val_score
score = cross_val_score(box_office_model, x, y_boxoffice, cv=5, scoring='r2')
print(f'Cross-validated R^2 scores: {score}')
print(f'Average R^2 score: {score.mean()}')
```

```python
import joblib
joblib.dump(box_office_model, 'box_office_prediction_model.pkl')
```

```python
#success_model

x_train, x_test, y_train_success, y_test_success = train_test_split(x, y_success, test_size=0.2, random_state=42)

from xgboost import XGBClassifier
from sklearn.metrics import classification_report

success_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', XGBClassifier
     (n_estimators=300,
      max_depth=5,
      learning_rate=0.1,
      random_state=42))
])
success_model.fit(x_train, y_train_success)
y_pred_success = success_model.predict(x_test)
print(classification_report(y_test_success, y_pred_success))
```
